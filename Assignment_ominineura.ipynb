```python
from crewai import Agent, Task, Crew, Process
from litellm import completion
import os
import sqlite3
import smtplib
from email.mime.text import MIMEText
from datetime import datetime
from typing import List, Dict, Any

# Configuration

# Set your Groq API key
os.environ["GROQ_API_KEY"] = "PASTE YOU GROQ API KEY"
CRM_EMAIL = "priyanhsudhaked00@gmail.com"
EMAIL_PASSWORD = "PriyanhsuDhaked@123"

# Database setup
def init_db():
    conn = sqlite3.connect('leads.db')
    c = conn.cursor()
    c.execute('''CREATE TABLE IF NOT EXISTS leads
                 (id INTEGER PRIMARY KEY AUTOINCREMENT,
                 company_name TEXT, company_size TEXT,
                 industry TEXT, budget_range TEXT,
                 timeline TEXT, use_case TEXT,
                 score INTEGER, is_qualified INTEGER,
                 date_added TIMESTAMP)''')
    conn.commit()
    conn.close()

init_db()

# Email notification
def send_crm_notification(lead_data: Dict[str, Any], evaluation: Dict[str, Any]):
    msg = MIMEText(
        f"New Qualified Lead:\nCompany: {lead_data['company_name']}\n"
        f"Industry: {lead_data['industry']}\nBudget: {lead_data['budget_range']}\n"
        f"Timeline: {lead_data['timeline']}\nScore: {evaluation['score']}/100\n"
        f"Use Case: {lead_data['use_case']}\n\nEvaluation Reasons:\n- " +
        "\n- ".join(evaluation['reasons'])
    )
    msg['Subject'] = f"New Qualified Lead: {lead_data['company_name']}"
    msg['From'] = CRM_EMAIL
    msg['To'] = CRM_EMAIL

    with smtplib.SMTP('smtp.gmail.com', 587) as server:
        server.starttls()
        server.login(CRM_EMAIL, EMAIL_PASSWORD)
        server.send_message(msg)

# Enhanced scoring algorithm
def enhanced_scoring(lead_data: Dict[str, Any]) -> Dict[str, Any]:
    score = 0
    reasons = []

    # Budget scoring
    if "100000" in lead_data['budget_range']:
        score += 40
        reasons.append("Strong budget allocation")
    elif "50000" in lead_data['budget_range']:
        score += 25
    else:
        score += 10

    # Company size scoring
    if "500" in lead_data['company_size']:
        score += 30
        reasons.append("Ideal company size")
    elif "100" in lead_data['company_size']:
        score += 20
    else:
        score += 10

    # Timeline scoring
    if "6" in lead_data['timeline']:
        score += 20
        reasons.append("Realistic timeline")
    elif "3" in lead_data['timeline']:
        score += 30
    else:
        score += 10

    # Industry bonus
    priority_industries = ["technology", "information", "software", "it"]
    if any(industry in lead_data['industry'].lower() for industry in priority_industries):
        score += 20
        reasons.append("Priority industry match")

    score = min(100, score)
    is_qualified = score >= 60

    return {
        "score": score,
        "is_qualified": is_qualified,
        "reasons": reasons
    }

# Custom Groq LLM integration
class GroqLLM:
    def __init__(self, model="groq/llama3-70b-8192"):
        self.model = model

    def generate_response(self, prompt):
        try:
            response = completion(
                model=self.model,
                messages=[{"content": prompt, "role": "user"}],
                api_key=os.environ["GROQ_API_KEY"]
            )
            return response.choices[0].message.content
        except Exception as e:
            print(f"LLM Error: {str(e)}")
            raise

# Initialize agents
groq_llm = GroqLLM()

information_collector = Agent(
    role="Lead Information Collector",
    goal="Gather complete lead details",
    backstory="Specialized in extracting key information from potential customers",
    llm=groq_llm,
    verbose=True
)

lead_evaluator = Agent(
    role="Lead Quality Evaluator",
    goal="Assess lead quality using advanced scoring",
    backstory="Expert in analyzing lead potential based on multiple factors",
    llm=groq_llm,
    verbose=True
)

lead_router = Agent(
    role="Lead Distribution Manager",
    goal="Route qualified leads to appropriate channels",
    backstory="Responsible for ensuring leads reach the right sales teams",
    llm=groq_llm,
    verbose=True
)

# Define tasks
def store_lead(lead_data: Dict[str, Any], evaluation: Dict[str, Any]):
    conn = sqlite3.connect('leads.db')
    c = conn.cursor()
    c.execute('''INSERT INTO leads VALUES
              (NULL,?,?,?,?,?,?,?,?,?)''',
              (lead_data['company_name'],
               lead_data['company_size'],
               lead_data['industry'],
               lead_data['budget_range'],
               lead_data['timeline'],
               lead_data['use_case'],
               evaluation['score'],
               int(evaluation['is_qualified']),
               datetime.now()))
    conn.commit()
    conn.close()

collect_info_task = Task(
    description="Collect company details including name, size, industry, budget, timeline, and use case.",
    expected_output="Complete lead information in structured format.",
    agent=information_collector,
    async_execution=False
)

evaluate_lead_task = Task(
    description="Evaluate lead using our enhanced scoring criteria.",
    expected_output="Lead evaluation with score and qualification status.",
    agent=lead_evaluator,
    context=[collect_info_task],
    function=enhanced_scoring
)

route_lead_task = Task(
    description="Route qualified leads to sales team and store in CRM.",
    expected_output="Confirmation of lead routing and storage.",
    agent=lead_router,
    context=[evaluate_lead_task],
    function=lambda ctx: {
        "store": store_lead(ctx[0], ctx[1]),
        "notify": send_crm_notification(ctx[0], ctx[1]) if ctx[1]['is_qualified'] else None
    }
)

# Create and run crew
lead_qualification_crew = Crew(
    agents=[information_collector, lead_evaluator, lead_router],
    tasks=[collect_info_task, evaluate_lead_task, route_lead_task],
    process=Process.sequential,
    verbose=True
)

# Execute
if __name__ == "__main__":
    try:
        result = lead_qualification_crew.kickoff()
        print("\nFinal Qualification Result:")
        print(result)
    except Exception as e:
        print(f"Execution Error: {str(e)}")
```
